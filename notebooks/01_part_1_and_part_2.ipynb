```python
import pandas as pd
import seaborn as sns
```

```python
insurance_data = pd.read_csv("../data/insurance.csv")
```

```python
insurance_data
```

```python
# We can find insights by plotting the data.
sns.scatterplot(
    x=insurance_data["bmi"], y=insurance_data["charges"], hue=insurance_data["smoker"]
)
```

```python
x = insurance_data.drop(columns=["region", "charges"])
y = insurance_data["charges"]

# Encoding or mapping categorical variables.
x["sex"] = x["sex"].map({"male": 1, "female": 0})
x["smoker"] = x["smoker"].map({"yes": 1, "no": 0})
```

```python
x.head()
```

```python
# Split the data into training and test sets.
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=42
)
```

```python
x_test.head()
```

```python
# Train the model.
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(x_train, y_train)
```

```python
# Predict values.
y_pred = model.predict(x_test)
```

```python
y_pred
```

```python
y_test
```

```python
# Evaluate the model.
from sklearn.metrics import r2_score

r2 = r2_score(y_test, y_pred)
print("R-squared:", r2)
```

```python
# Calculate Adjusted R-squared.
x_test.shape
n = x_test.shape[0]
p = x_test.shape[1]

adjusted_r2 = 1 - ((1 - r2) * (n - 1) / (n - p - 1))
print("Adjusted R-squared:", adjusted_r2)
```

```python
# Add interaction features.
x = insurance_data.drop(columns=["charges"])
y = insurance_data["charges"]

x = pd.get_dummies(x, columns=["region"], drop_first=False, dtype=int)

# Encoding or mapping categorical variables.
x["sex"] = x["sex"].map({"male": 1, "female": 0})
x["smoker"] = x["smoker"].map({"yes": 1, "no": 0})

x["age_smoker"] = x["age"] * x["smoker"]
x["bmi_smoker"] = x["bmi"] * x["smoker"]
```

```python
x.head()
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=42
)
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(x_train, y_train)
```

```python
y_pred = model.predict(x_test)
```

```python
# We see that the R-squared value has improved after adding interaction features.
from sklearn.metrics import r2_score

r2 = r2_score(y_test, y_pred)
print("R-squared:", r2)
```

```python
# Check for underfitting and overfitting.
# If R2 train is high and R2 test is low, then the model is overfitting.
# If R2 train is low and R2 test is low, then the model is underfitting.
y_train_pred = model.predict(x_train)
r2_train = r2_score(y_train, y_train_pred)

print("R-squared train:", r2_train)
print("R-squared test:", r2)
```
